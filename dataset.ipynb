```python
import sys
from path import PathFinder
PATHS = PathFinder()
sys.path.insert(0, PATHS["LIB_PYNICHE"].as_posix())
from pyniche.data.yolo.API import YOLO_API
yolo_top = YOLO_API(PATHS["DIR_DATA"] / "1_top")
yolo_side = YOLO_API(PATHS["DIR_DATA"] / "2_side")
yolo_extVT = YOLO_API(PATHS["DIR_DATA"] / "e_VT")
yolo_extIMU = YOLO_API(PATHS["DIR_DATA"] / "e_IMU")
```

```python
# apply augmentation
image = yolo_side["daylight"].get_PIL(100)
from PIL import Image
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import random
# Define a transform pipeline with flip and color jitter
random.seed(0)
transform = transforms.Compose([
    transforms.RandomHorizontalFlip(p=0.8),  # Always flip
    transforms.ColorJitter(brightness=0.5, contrast=0.5, saturation=0.5),
    transforms.RandomAffine(degrees=0, translate=(0.3, 0.3), scale=(0.8, 1.2)),    
]
# Apply transform to generate two augmented images
augmented1 = transform(image)
augmented2 = transform(image)

# Show original and augmented images
fig, axs = plt.subplots(1, 3, figsize=(12, 4), dpi=300)
axs[0].imshow(image)
axs[0].set_title("Original")
axs[1].imshow(augmented1)
axs[1].set_title("Augmented 1")
axs[2].imshow(augmented2)
axs[2].set_title("Augmented 2")
for ax in axs:
    ax.axis('off')
plt.tight_layout()
# Save augmented images
plt.savefig("ms/r1/src/fig_aug.png", dpi=300)
```
